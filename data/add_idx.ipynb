```python
import pandas as pd

# Read both CSV files
vehicle_df = pd.read_csv('vehicle.csv')
carmodel_res_df = pd.read_csv('carmodel-res.csv')

# Make sure both dataframes have vehicle_id
vehicle_df['vehicle_id'] = vehicle_df.index
carmodel_res_df['vehicle_id'] = carmodel_res_df.index

# Add the model column from carmodel_res to vehicle dataframe
vehicle_df['model'] = carmodel_res_df['model']


# Save the updated vehicle dataframe back to CSV
# vehicle_df.to_csv('vehicle.csv', index=False)
```

```python
vehicle_df['Brand'] = vehicle_df['Brand'].str.replace('/', '-')
```

```python
import pandas as pd
import os

# # Read the vehicle CSV
# df = pd.read_csv('vehicle.csv')

# Function to create image URL based on brand and model
def get_image_url(brand, model):
    # Clean brand and model names (remove spaces, special characters if needed)
    brand = str(brand).strip().replace(' ', '_')
    model = str(model).strip().replace(' ', '_')
    
    # Create the image URL
    image_name = f"{brand}_{model}.jpg"  # or whatever your image extension is
    image_url = f"/static/car_images/{image_name}"
    
    # Optionally verify if the image exists
    if os.path.exists(f"static/car_images/{image_name}"):
        return image_url
    else:
        return "/static/car_images/default-car.jpg"  # Return default image URL if specific image doesn't exist

# Add image_url column
vehicle_df['image_url'] = vehicle_df.apply(lambda row: get_image_url(row['Brand'], row['model']), axis=1)

# Save the updated dataframe back to CSV
vehicle_df.to_csv('vehicle.csv', index=False)
```
